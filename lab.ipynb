```python
# 创建一个函数来合并问题和回答
def merge_qa(example):
    text = f"{example['question']}\n\n{example['responses'][0]['response']}"
    return {"text": text}

# 应用函数创建新字段
new_ds = ds.map(merge_qa)

# 如果您只想保留 text 字段，可以选择需要的列
new_ds = new_ds.select_columns(["text"])

# 查看结果
print(new_ds[0])

# 保存新数据集（可选）
new_ds.save_to_disk("merged_qa_dataset")

# 或者导出为Hugging Face数据集（可选）
# new_ds.push_to_hub("your-username/merged-qa-dataset")
```
